```python
import pandas as pd
dp=pd.read_csv("/content/preprocessing_practice_dataset.csv")
```

encoding

```python
from sklearn.preprocessing import LabelEncoder

encoded_data=dp.copy()

one_hot_encoded = pd.get_dummies(encoded_data['Education'], prefix='Education')
encoded_data = pd.concat([encoded_data, one_hot_encoded], axis=1)

# Since 'Pclass' is ordinal, we can create a mapped column for demonstration.
education_mapping = {1: 'Masters', 2: 'Bachelors', 3: 'phD'}
# 3. Ordinal Encoding For p class
encoded_data['education_Category'] = encoded_data['Education'].map(education_mapping)

# Display the resulting dataset with relevant encoded columns
print(encoded_data[['Gender', 'Department', 'Education', 'Education_Masters', 'Education_Bachelors', 'Education_PhD', 'City']].head())
```

scaling

```python
import pandas as pd
#skikitlearn
from sklearn.preprocessing import MinMaxScaler,StandardScaler,RobustScaler
```

```python
practice_data=pd.read_csv("/content/preprocessing_practice_dataset.csv")


#nnumerical columns
numerical_col=['Age','Performance_Rating']

#drop rows with missing values
practice_data.dropna(subset=numerical_col,inplace=True)

#initilize scalers
minmax_scaler=MinMaxScaler()
standard_scaler=StandardScaler()
robust_scaler=RobustScaler()


#applying minmax scaling
practice_data['Age_minmax']=minmax_scaler.fit_transform(practice_data[['Age']])
practice_data['perfomance_minmax']=minmax_scaler.fit_transform(practice_data[['Performance_Rating']])


#apply standard scaling

practice_data['Age_standard']=standard_scaler.fit_transform(practice_data[['Age']])
practice_data['perfomance_standard']=standard_scaler.fit_transform(practice_data[['Performance_Rating']])


#apply robust scaling

practice_data['Age_robust']=robust_scaler.fit_transform(practice_data[['Age']])
practice_data['perfomance_robust']=robust_scaler.fit_transform(practice_data[['Performance_Rating']])
practice_data[['Age','Age_minmax','perfomance_minmax','Age_standard','perfomance_standard','Age_robust','perfomance_standard']]

```
