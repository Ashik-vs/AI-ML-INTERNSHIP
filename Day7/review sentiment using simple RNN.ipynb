```python
import pandas as pd
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.layers import Embedding,SimpleRNN,Dense

vocab_size=10000

#keep only the to 10000 most frequent words

(x_train,y_train),(x_test,y_test)=imdb.load_data(num_words=vocab_size)

print("training samples :",len(x_train))
print("testing samples :",len(x_test))
```

```python
max_length=200
x_train=pad_sequences(x_train,
                      maxlen=max_length,
                      padding='post')
x_test=pad_sequences(x_test,
                     maxlen=max_length,
                     padding='post')

model=Sequential([
    Embedding(input_dim=vocab_size,output_dim=64,input_length=max_length),
    SimpleRNN(64),
    Dense(32,activation='relu'),
    Dense(1,activation='sigmoid')
])
```

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
model.summary(

)
```

```python
history=model.fit(
    x_train,y_train,
    epochs=5,
    batch_size=64,
    validation_split=0.2
)
```

```python
loss,accuracy=model.evaluate(x_test,y_test)
print("accuracy:",accuracy)
```

```python
from tensorflow.keras.datasets import imdb
word_index=imdb.get_word_index()
#shift indices by 3 cuz keras reserves 0,1,2

word_index={k:(v+3) for k,v in word_index.items()}
word_index["<PAD>"]=0
word_index["<START>"]=1
word_index["<UNK>"]=2
word_index["<UNUSED>"]=3
```

```python
def review_to_sequence(review):
  review=review.lower().split()
  sequence=[]
  for word in review:
    sequence.append(word_index.get(word,2))
  return sequence

```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
import numpy as np

max_length=200
def predict_sentement(review):
  sequence=review_to_sequence(review)
  padded=pad_sequences(
      [sequence],
      maxlen=max_length,
      padding='post'
  )
  prediction=model.predict(padded,verbose=0)
  score=prediction[0][0]

  print("sentiment score:",score)

  if score>0.5:
    print("positive review")
  else:
    print("negative review")

```

```python
predict_sentement("this is a good movie")
```
